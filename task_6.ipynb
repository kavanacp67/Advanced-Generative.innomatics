```python
def manage_parking(capacity, logs):
    parked = 0

    for action in logs:
        if action == "IN":
            parked += 1
        elif action == "OUT" and parked > 0:
            parked -= 1

    print("Currently Parked Vehicles:", parked)

    if parked > capacity:
        print("Parking Status: Exceeded")
    else:
        print("Parking Status: Available")


# Sample Input
manage_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
def estimate_delivery_time(distance, traffic, weather):
    time = distance * 5

    if traffic.lower() == "high":
        time += 10
    elif traffic.lower() == "medium":
        time += 5

    if weather.lower() == "rainy":
        time += 5
    elif weather.lower() == "stormy":
        time += 15

    print("Estimated Delivery Time:", time, "minutes")


# Sample Input
estimate_delivery_time(8, "High", "Rainy")
```

```python
def analyze_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = int((booked_count / total_seats) * 100)

    print("Occupancy:", str(occupancy) + "%")

    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")


# Sample Input
analyze_occupancy(200, [1] * 150)
```

```python
def classify_server_load(cpu_readings):
    average = int(sum(cpu_readings) / len(cpu_readings))

    print("Average CPU Load:", str(average) + "%")

    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Input
classify_server_load([45, 60, 70, 85, 90])
```

```python
def monitor_resources(resource_usage):
    overused = []

    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
monitor_resources({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

```python
def manage_event_registration(capacity, registrations):
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Input
manage_event_registration(100, 105)
```
